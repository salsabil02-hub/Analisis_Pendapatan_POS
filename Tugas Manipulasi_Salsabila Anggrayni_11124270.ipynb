```python
import pandas as pd
import numpy as np

# Membaca dataset
df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')

# Menampilkan 5 data pertama
display(df.head())

# Melihat informasi dataset
df.info()
```

Task 1 — Indexing & Selection

```python
# 1.1 Pembersihan format harga: koma -> titik, lalu jadi float
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace(',', '.', regex=False).astype(float)
print(df['harga_satuan'].dtype)
```

```python
# 1.2 Ubah jumlah negatif menjadi positif
print('Negatif sebelum:', (df['jumlah'] < 0).sum())
df.loc[df['jumlah'] < 0, 'jumlah'] = df.loc[df['jumlah'] < 0, 'jumlah'].abs()
print('Negatif sesudah:', (df['jumlah'] < 0).sum())
```

```python
# 1.3 Isi NaN kode_kasir dengan 'K_UNKNOWN'
print('NaN sebelum:', df['kode_kasir'].isna().sum())
df.loc[df['kode_kasir'].isna(), 'kode_kasir'] = 'K_UNKNOWN'
print('NaN sesudah:', df['kode_kasir'].isna().sum())
```

```python
# 1.4 5 baris pertama, kolom id_transaksi, tanggal, kode_gerai, jumlah
subset_iloc = df.iloc[:5, [0, 1, 2, 5]]
subset_iloc
```

Task 2 — Operasi Aritmatika

```python
# 2.1 Total_Kotor
df['Total_Kotor'] = df['harga_satuan'] * df['jumlah']

# 2.2 Nominal_Diskon (diskon_persen dibagi 100 dulu)
df['Nominal_Diskon'] = df['Total_Kotor'] * (df['diskon_persen'] / 100)

# 2.3 Total_Bersih (setelah diskon + PPN 11%)
df['Total_Bersih'] = (df['Total_Kotor'] - df['Nominal_Diskon']) * 1.11

df[['id_transaksi', 'Total_Kotor', 'Nominal_Diskon', 'Total_Bersih']].head()
```

Task 3 — Filtering

```python
# 3.1 SELESAI, gerai G05 atau G09, Total_Bersih > 100.000
filter_multi = df[
    (df['status'] == 'SELESAI') &
    (df['kode_gerai'].isin(['G05', 'G09'])) &
    (df['Total_Bersih'] > 100000)]
print('Jumlah baris:', len(filter_multi))   
filter_multi.head()
```

```python
# 3.2 query(): gerai G10 dan jumlah >= 5
filter_g10 = df.query('kode_gerai == "G10" and jumlah >= 5')
print('Jumlah baris:', len(filter_g10))     # hasil: 39
filter_g10.head()
```

Task 4 — Sorting

```python
# 4.1 kode_gerai A-Z, lalu Total_Bersih tertinggi ke terendah
df_sorted = df.sort_values(by=['kode_gerai', 'Total_Bersih'], ascending=[True, False])
df_sorted.head(10)
```

```python
# 4.2 Top 5 transaksi dengan Total_Bersih tertinggi
top_5_transaksi = df.nlargest(5, 'Total_Bersih')
top_5_transaksi
```

Task 5 — Grouping & Agregasi

```python
# 5.1 Laporan per gerai (hanya status SELESAI)
df_selesai = df[df['status'] == 'SELESAI']

laporan_gerai = df_selesai.groupby('kode_gerai', as_index=False).agg(
    Jumlah_Transaksi=('id_transaksi', 'count'),
    Total_Unit_Terjual=('jumlah', 'sum'),
    Total_Omzet_Bersih=('Total_Bersih', 'sum'),
    Rata_Rata_Transaksi=('Total_Bersih', 'mean')
).sort_values(by='Total_Omzet_Bersih', ascending=False).round(2)

laporan_gerai
```

```python
# 5.2 Persentase kontribusi tiap transaksi terhadap omzet gerainya
total_per_gerai = df.groupby('kode_gerai')['Total_Bersih'].transform('sum')
df['Kontribusi_Gerai_Persen'] = df['Total_Bersih'] / total_per_gerai * 100

df[['id_transaksi', 'kode_gerai', 'Total_Bersih', 'Kontribusi_Gerai_Persen']].head(10)
```
